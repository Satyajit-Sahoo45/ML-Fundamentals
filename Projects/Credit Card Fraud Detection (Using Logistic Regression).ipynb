```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
dataset = pd.read_csv(r'../data/creditcard.csv')
dataset.head(4)
```

```python
dataset.info()
```

```python
dataset.isnull().sum()
```

```python
dataset['Class'].value_counts()
```

```python
# 0 → Normal Transaction

# 1 → Fraudulent transaction
```

```python
# separating the data for analysis

legit = dataset[dataset.Class == 0]
fraud = dataset[dataset.Class == 1]
```

```python
# Statistical measures of data

legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
# compare the values of both 
dataset.groupby('Class').mean()
```

```python
# select randomly 492 instances from legit transaction for equal distribution of both classes.
# it will balanced dataset suitable for training our fraud detection model.
# balancing the dataset enhances the models's accuracy

legit_sample = legit.sample(n=492)
```

```python
new_dataset = pd.concat([legit_sample, fraud], axis=0)
new_dataset.head(4)
```

```python
new_dataset.columns
```

```python
new_dataset.groupby("Class").mean()
```

```python
X = new_dataset.iloc[::-1]
Y = new_dataset["Class"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
print(X_train.shape, X_train.shape, X_test.shape)
```

```python
model = LogisticRegression()
```

```python
# training the Logistic Regression modle with training data

model.fit(X_train, Y_train)
```

```python
model.score(X_train, Y_train) * 100
```

```python
# accuracy on training data

training_data_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(training_data_prediction, Y_train)
training_data_accuracy * 100
```

```python
# indicates that the model correctly predicts the class labels for arround 95.42% of the instaces in the training data.
```

```python
# accuracy of training ata

test_data_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(test_data_prediction, Y_test)
test_data_accuracy * 100
```

```python
# the accuracy score the test data is approximately 93.40%.
# this indicates the LR-model correctly predicts the class lables for around 93.40% of the instances in the testing data.
```
