```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset = pd.read_csv(r"./data/placement-dataset(Multi Logistic Regression).csv")
dataset.head(4)
```

```python
# to draw a distribution plot
# sns.kdeplot(data=dataset["iq"])
# plt.show()
```

```python
dataset.isnull().sum()
```

```python
sns.scatterplot( x="cgpa", y="iq", data=dataset, hue="placement")
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["placement"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=2, test_size=0.2)
```

# Naive Bayes(Gaussian NB, Multinomial NB, Bernoulli NB)

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
```

```python
gnb = GaussianNB()
gnb.fit(x_train, y_train)
```

```python
gnb.score(x_test, y_test) * 100, gnb.score(x_train, y_train) * 100
```

```python
mnb = MultinomialNB()
mnb.fit(x_train, y_train)
```

```python
mnb.score(x_test, y_test) * 100, mnb.score(x_train, y_train) * 100
```

```python
bnb = BernoulliNB()
bnb.fit(x_train, y_train)
```

```python
bnb.score(x_test, y_test) * 100, bnb.score(x_train, y_train) * 100
```

```python
# to see decision regions of each model
```

```python
plot_decision_regions(x.to_numpy(), y.to_numpy(), clf=gnb)
plt.show()
```

```python
plot_decision_regions(x.to_numpy(), y.to_numpy(), clf=mnb)
plt.show()
```

```python
plot_decision_regions(x.to_numpy(), y.to_numpy(), clf=bnb)
plt.show()
```
