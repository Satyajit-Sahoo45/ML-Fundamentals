```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset = pd.read_csv(r"C:\Users\user\Documents\jnotebook\data\multiple_linear_regression_dataset.csv")
```

```python
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["salary"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=2, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree
```

```python
dt = DecisionTreeRegressor()
dt.fit(x_train, y_train)
```

```python
dt.score(x_test, y_test) * 100
```

```python
plt.figure(figsize=(50,50))
plot_tree(dt)
plt.show()
```

# Check Over-Fitting

```python
dt.score(x_train, y_train) * 100
```

```python
# our test accuracy score(84.299) and train data accuracy score(100.0) seems to have large difference.
# If there is a difference beetween train score and test score then we might have Over-Fitting
```
