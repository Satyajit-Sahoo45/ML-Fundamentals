```python
import pandas as pd
```

```python
dataset = pd.read_csv(r"./data/Social_Network_Ads.csv")
dataset.head(4)
```

```python
x = dataset.iloc[:, :-1]
y = dataset["Purchased"]
```

```python
dataset["Purchased"].value_counts()
```

# using Random Under Sampling

```python
from imblearn.under_sampling import RandomUnderSampler
```

```python
ru = RandomUnderSampler()
```

```python
ru_x, ru_y = ru.fit_resample(x,y)
```

```python
# to check if the '0' become same no. as '1' after sampling.

ru_y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(ru_x, ru_y, random_state=2, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
lr.score(x_test, y_test) * 100
```

```python
lr.predict([[46, 28000]])
```
