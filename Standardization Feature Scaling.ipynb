```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"/home/satyajit-sahoo/Downloads/bank-loan.csv")
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
dataset["income"] = dataset["income"].fillna(dataset["income"].mean())
```

```python
sns.distplot(dataset["income"])
plt.show()
```

```python
dataset.describe()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
ss.fit(dataset[["income"]])
```

```python
dataset["income_ss"] = pd.DataFrame(ss.transform(dataset[["income"]]), columns=["x"])
```

```python
dataset.head(4)
```

```python
dataset.describe()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(dataset["income"])

plt.subplot(1,2,2)

plt.title("After")
sns.distplot(dataset["income_ss"])
plt.show()
```
