```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset = pd.read_csv(r"./data/placement-dataset(Multi Logistic Regression).csv")
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
sns.scatterplot(x="cgpa", y="iq", data=dataset, hue="placement")
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["placement"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=4)
```

```python
from sklearn.svm import SVC
```

```python
# linear, poly, rbf, sigmoid, precomputed

svc = SVC(kernel="linear")
svc.fit(x_train, y_train)
```

```python
svc.score(x_test, y_test) * 100
```

```python
svc.score(x_train, y_train) * 100
```

```python
plot_decision_regions(x.to_numpy(), y.to_numpy(), clf=svc)
plt.show()
```
