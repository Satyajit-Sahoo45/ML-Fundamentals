```python
import pandas as pd

dataset = pd.read_csv(r"/home/satyajit-sahoo/Downloads/bank-loan.csv")
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
dataset["Gender"] = dataset["Gender"].fillna(dataset["Gender"].mode()[0])
```

```python
en_data = dataset[["Gender", "married"]]
en_data
```

```python
# 1. one hot encoding using get-dummies
# the data's shows as boolean type

pd.get_dummies(en_data).info()
```

```python
# 2. to get results in binary type using scikit-learn

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop="first")
ar = ohe.fit_transform(en_data).toarray()
```

```python
pd.DataFrame(ar,columns=["Gender_Male","married_Yes"])
```
