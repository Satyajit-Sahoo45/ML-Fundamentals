```python
import pandas as pd
df = pd.DataFrame({"size": ["s", "m", "l", "xl", "s", "m", "l", "s", "s", "l", "xl", "m"]})
df.head(4)
```

```python
ord_data = [["s", "m", "l", "xl"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=ord_data)
```

```python
oe.fit(df[["size"]])
```

```python
df["en_size"] = oe.transform(df[["size"]])
```

```python
df
```

```python
# encoding using map function
```

```python
ord_data1 = {"s": 0, "m": 1, "l": 2, "xl": 3}
df["size_en_map"] = df["size"].map(ord_data1)
```

```python
df
```

```python
dataset = pd.read_csv(r"/home/satyajit-sahoo/Downloads/bank-loan.csv")
dataset["area"] = dataset["area"].fillna(dataset["area"].mode()[0])
```

```python
en_data_ord = [['Rural', "Urban"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder

oen = OrdinalEncoder(categories=en_data_ord)
```

```python
oen
```

```python
dataset["area"] = oen.fit_transform(dataset[["area"]])
```
