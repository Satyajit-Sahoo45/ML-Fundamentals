```python
import pandas as pd
import numpy as np
```

```python
dataset = pd.read_csv(r"./data/groceries.csv")
dataset.head(3)
```

```python
dataset.shape
```

```python
market = []

for i in range(0, dataset.shape[0]):
    cus = []
    for j in dataset.columns:
        if type(dataset[j][i]) == str:
            cus.append(dataset[j][i])
    market.append(cus)
```

```python
l = []
for i in market:
    for j in i:
        l.append(j)
```

```python
import collections
```

```python
p = collections.Counter(l)
```

```python
d = {"Item Name": p.keys(), "values": p.values()}
```
