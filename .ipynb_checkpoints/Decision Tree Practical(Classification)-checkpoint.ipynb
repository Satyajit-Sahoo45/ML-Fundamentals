```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\user\Documents\jnotebook\data\Social_Network_Ads.csv")
```

```python
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["Purchased"]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
ss.fit(x)
```

```python
x = pd.DataFrame(ss.transform(x), columns=x.columns)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=2, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# It used GINI index for tree splitting by default...to use "Entropy"
dtc = DecisionTreeClassifier()
```

```python
dtc.fit(x_train, y_train)
```

```python
dtc.score(x_test, y_test) * 100
```

```python
dtc.predict([[35, 20000]])
```

# Draw Dicision Tree(Root Node, Sub Nodes)

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(50,50))
plot_tree(dtc)
# plt.savefig("demo.jpg")
plt.show()
```
