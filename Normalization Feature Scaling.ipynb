```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"/home/satyajit-sahoo/Downloads/bank-loan.csv")
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
dataset.describe()
```

```python
sns.distplot(dataset["income"])
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
ms = MinMaxScaler()
ms.fit(dataset[["income"]])
```

```python
dataset["income_mms"] = ms.transform(dataset[["income"]])
```

```python
dataset.head(4)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1, 2, 1)
plt.title("Before")
sns.distplot(dataset["income"])

plt.subplot(1, 2, 2)
plt.title("After")
sns.distplot(dataset["income_mms"])
plt.show()
```
